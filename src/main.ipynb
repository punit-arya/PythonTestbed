```python
def toEnggNotation(n, scientificNotation = True, nSignificantDigits = 4):
	exp = 0
	oldN = n
	if n > 1:
		while n >= 1000:
			exp += 3
			n /= 1000
		prefixes = "", "k", "M", "G", "T", "P", "E"
		return f"{n:.3f}{prefixes[exp//3]}\t{oldN}" if scientificNotation else f"{n}{prefixes[exp//3]}"
	elif n < 1:
		while n < 1:
			exp -= 3
			n *= 1000
		prefixes = "", "m", "\N{GREEK SMALL LETTER MU}", "n", "p", "f", "a"
		return f"{n:.{nSignificantDigits}f}{prefixes[abs(exp)//3]}\t{oldN}" if scientificNotation else f"{n}{prefixes[abs(exp)//3]}"
	else:
		return 1
```

```python
def getExtremum(expr, interval, resolution, minimax):

	def f(expr):

		def g(x):
			return eval(expr)

		return g

	evalExpr = f(expr)
	l = [interval[0] + (i * (interval[1] - interval[0]) / (resolution - 1)) for i in range(resolution)]  # Cheated with the calc.
	return min(map(evalExpr, l)) if minimax == "minimum" else max(map(evalExpr, l))


expr = "x ** 2 - 1"
print(getExtremum(expr, (-10, 10), 5, "minimum"))
print(getExtremum(expr, (-10, 10), 5, "maximum"))
print(getExtremum(expr, (-10, 10), 6, "minimum"))
print(getExtremum(expr, (-10, 10), 6, "maximum"))
expr = "math.log10(x)"
print(getExtremum(expr, (1, 100), 20, "minimum"))
print(getExtremum(expr, (1, 100), 20, "maximum"))
expr = "math.sin(x)"
print(getExtremum(expr, (-10, 10), 20, "minimum"))
print(getExtremum(expr, (-10, 10), 20000, "maximum"))
expr = "x"
print(getExtremum(expr, (1, 6), 8, "minimum"))
print(getExtremum(expr, (1, 6), 8, "maximum"))
```

```python
def power(x, y):
	return x ** y


def dist(pt1, pt2):
	return math.sqrt(sum(coord1 - coord2 for coord1, coord2 in zip(pt1, pt2)))


def partial(f, operand2):

	def calcPartialFunction(operand1):
		return f(operand1, operand2)

	return calcPartialFunction


squares = partial(power, 2)
print(squares(3))
print(squares(4))
dist_from_origin = partial(dist, (0, 0))
print(dist_from_origin((1, 1)))
gcd_13 = partial(math.gcd, 13)
log_2 = partial(math.log, 2)
log_10 = partial(math.log, 10)
log_16 = partial(math.log, 16)
print(log_2(8))
print(log_10(100))
print(log_16(65536))
```

```python
def norm(x, y):
	return math.sqrt(x ** 2 + y ** 2)


def find_index_min(seq):
	minima = min(seq)
	return seq.index(minima)


def logged(f):

	def g(*args):
		print("Logging...")
		start = time.perf_counter()
		print(f(*args))
		end = time.perf_counter()
		print(toEnggNotation(end - start))
		print("... Finished")

	return g


loggedNorm = logged(norm)
loggedNorm(3, 4)
loggedFindIndexMin = logged(find_index_min)
loggedFindIndexMin([8, 7, 2, 9, 4])
```

```python
def encloseGetPower(f):

	def getPower(*args):
		return f(*args)

	print(hex(id(getPower)))
	return getPower


getPower = encloseGetPower(pow)
print(getPower(2, 5))
print(hex(id(getPower)))
print(hex(id(encloseGetPower)))
print(getPower.__closure__)
print(encloseGetPower.__closure__)
```

```python
suits = 'shdc'  # Spades, Hearts, Diamonds, Clubs
ranks = list('23456789') + ['10', 'J', 'Q', 'K', 'A']
cards = [[rank + suit for rank in ranks] for suit in suits]


def sorter(suits, reverse = False):
	return sorted(suits, key = lambda card: ranks.index(card.strip("s").strip("h").strip("d").strip("c")), reverse = reverse)


sortedCards = []
for suit in cards:
	sortedCards.append(sorter(suit, reverse = True))
print(sortedCards)

sortedCards = []
for suit in cards:
	sortedCards.append(sorter(suit))
print(sortedCards)
```
