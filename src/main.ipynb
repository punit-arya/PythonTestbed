```python
total, count = (0,) * 2
data = [10.5, 11.2, 9.8, None, 11.5, None]
for e in data:
	if e:
		total += e
		count += 1
average = total / count
print(average)
```

```python
for i, e in enumerate(data):
	if not e:
		data[i] = average
print(data)
```

```python
import statistics


data = [10.5, 11.2, 9.8, None, 11.5, None]
average = statistics.fmean(e for e in data if e)
data = [average if e is None else e for e in data]
print(data)
```
