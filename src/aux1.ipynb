#### LIST OF SERIES (TRANSPOSE NEEDED)

```python
boroughs = pd.Index(["The Bronx", "Brooklyn", "Manhattan", "Queens", "Staten Island"])
counties = pd.Series(["Bronx", "Kings", "New York", "Queens", "Richmond"], index = boroughs, name = "County")
populations = pd.Series([1_000_000, 2_000_000, 3_000_000, 4_000_000, 5_000_000], index = boroughs, name = "Population")
gdp = pd.Series([1.00, 1.01, 2.02, 3.03, 4.04], index = boroughs, name = "GDP")
areas = pd.Series([2000, 3000, 4000, 5000, 6000], index = boroughs, name = "Area")
newYork = [counties, populations, gdp, areas]
df = pd.DataFrame(newYork)
print(df.transpose())
```

```python
def returnHighestVolume(df, symbol):
	try:
		stocks = df[df.index == symbol]
		stocks = stocks.sort_values("volume", ascending = False)
		highestVolume = stocks["volume"].iloc[0]
		return stocks[stocks["volume"] == highestVolume]
	except IndexError as ex:
		raise ValueError("symbol not found")


try:
	print(returnHighestVolume(quotes, "MSFT"))
except ValueError as ex:
	print("ValueError!")
```

```python
def returnLargestSpread(df, symbol):
	stocks = df[df.index == symbol]
	spread = stocks.high - stocks.low
	stocks = pd.concat([stocks, spread], axis = 1).rename(columns = {0: "spread"}).sort_values("spread", ascending = False)
	return stocks[:1]


print(returnLargestSpread(quotes, "MSFT"))
```

```python
def returnLargestSpread(df, symbol):
	stocks = df[df.index == symbol]
	spread = stocks.high - stocks.low
	stocks = pd.concat([stocks, spread], axis = 1).rename(columns = {0: "spread"}).sort_values("spread", ascending = False)
	return stocks[:1]


symbols = quotes.index.unique()
print(symbols)
for symbol in symbols:
	print(returnLargestSpread(quotes, symbol))
```
