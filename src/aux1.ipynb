```python
with open("../var/data.csv") as fh:
	data = csv.reader(fh)
	next(data)
	data = np.array(list(data)).astype(np.float32)
	# velocities = np.array([0.0] * 99)
	# for i in range(1, 100):
	# 	velocities[i - 1] = (data[i, 1] - data[i - 1, 1]) / (data[i, 0] - data[i - 1, 0])
	velocities = (data[1:, 1] - data[:-1, 1]) / (data[1:, 0] - data[:-1, 0])
	print(velocities)
	print(velocities.size)
	minima = np.min(velocities)
	maxima = np.max(velocities)
	average = np.mean(velocities)
	stddev = np.std(velocities)
	print(minima, maxima, average, stddev)
```
